## Database - F418497 wrote on 11/12/2024

```python
'''
Converts .txt files into list of lists

See Rental.txt, Music_Info.txt and Music_Feedback.txt for how the csv format
of each is converted into a list for each line of text and separates items
by commas.

Files:

    Music_Info.txt
    Rental.txt
    Music_Feedback.txt

Misc variables:

    MusicList
    RentalList
    FeedbackList
'''
#reads Music_Info.txt into a list of lists called MusicList
MusicFile = open("Music_Info.txt","r") 
MusicList = []
for line in MusicFile:
    MusicString = line.strip()
    MusicStrings = MusicString.split(",")
    MusicList.append(MusicStrings)
MusicFile.close()
#reads Rental.txt into a list of lists called RentalList
RentalFile = open("Rental.txt","r")
RentalList = []
for line in RentalFile:
    RentalString = line.strip()
    RentalStrings = RentalString.split(",")
    RentalList.append(RentalStrings)
RentalFile.close()
#reads Music_Feedback into a list of lists called FeedbackList
FeedbackFile = open("Music_Feedback.txt","r")
FeedbackList = []
for line in FeedbackFile:
    FeedbackString = line.strip()
    FeedbackStrings = FeedbackString.split(",")
    FeedbackList.append(FeedbackStrings)
FeedbackFile.close()
```

## Music Search - F418497 wrote on 11/12/2024

```python
'''
Selects an input string and input dropdown of strings and returns a list of all the details
matching the string and type.

Functions:

    artistSearch(string) -> list
    titleSearch(string) -> list
    mediumSearch(string) -> list
    genreSearch(string) -> list
    searchReturn(string, string)
    onSearchChange(boolean)

Misc variables:

    searchList
    musicSearchBar
    typeDropdown
    searchOutput
'''
import ipywidgets as widgets
from IPython.display import clear_output
searchList = []
def artistSearch(musicString):
    '''
    Searches through artists and appends them to a list if the input string matches.

        Parameters:
            musicString (str): The string that is is inputted to match an artist.

        Returns:
            searchList (list): The list of artists that match the input string.
    '''
    searchList.clear()
    for j in range (len(MusicList)):
        if MusicList[j][1].strip().lower() == musicString.strip().lower():
            searchList.append(MusicList[j])
    return searchList
        
def titleSearch(musicString):
    '''
    Searches through titles and appends them to a list if the input string matches.

        Parameters:
            musicString (str): The string that is is inputted to match a title.

        Returns:
            searchList (list): The list of titles that match the input string.
    '''
    searchList.clear()
    for j in range (len(MusicList)):
        if MusicList[j][2].strip().lower() == musicString.strip().lower():
            searchList.append(MusicList[j])
    return searchList
def mediumSearch(musicString):
    '''
    Searches through mediums and appends them to a list if the input string matches.

        Parameters:
            musicString (str): The string that is is inputted to match a medium.

        Returns:
            searchList (list): The list of mediums that match the input string.
    '''
    searchList.clear()
    for j in range (len(MusicList)):
        if MusicList[j][3].strip().lower() == musicString.strip().lower():
            searchList.append(MusicList[j])
    return searchList

def genreSearch(musicString):
    '''
    Searches through genres and appends them to a list if the input string matches.

        Parameters:
            musicString (str): The string that is is inputted to match a genre.

        Returns:
            searchList (list): The list of genres that match the input string.
    '''
    searchList.clear()
    for j in range (len(MusicList)):
        if MusicList[j][4].strip().lower() == musicString.strip().lower():
            searchList.append(MusicList[j])
    return searchList
# create music searchbar
musicSearchBar = widgets.Text( 
    value='',
    placeholder='Search for music',
    description='Search:',
    disabled = False,
)
musicSearchBar.style.text_color = 'teal'
# create type dropdown menu
typeDropdown = widgets.Dropdown(
    options=['Artist', 'Title', 'Medium', 'Genre'],
    value='Artist',
    description='Type:',
)
typeDropdown.style.text_color = 'teal'
# create search output
searchOutput = widgets.Output(layout=widgets.Layout(height='150px', width='700px'))
# return type of searched string depending on type picked
def searchReturn(musicString, searchTypes): 
    '''
    Search for the inputted string and type of search to enter the respective function for the type chosen.

        Parameters:
            musicString (str):The inputted string for the name of artist/title/medium/genre
            searchTypes (str):The inputted string for the type of search
    '''
    if searchType == "Artist":
        artistSearch(musicString)
        return artistSearch(musicString)
    elif searchType == "Title":
        titleSearch(musicString)
        return titleSearch(musicString)
    elif searchType == "Medium":
        mediumSearch(musicString)
        return mediumSearch(musicString)
    else:
        genreSearch(musicString)
        return genreSearch(musicString)
# when there is a change in the searchbar, update values and clear previous output
def onSearchChange(change): 
    '''
    Updates the output everytime the input is changed.

        Parameters:
            change(bool):The boolean variable that calls the function when it is true
    '''
    global searchType
    searchType = typeDropdown.value
    searchString = musicSearchBar.value
    with searchOutput:
        clear_output()
        if searchReturn(searchString, searchType) == [] or searchReturn(searchString, searchType) == None:
            print("There are no songs available.")
        else:
            print("| Record ID | Artist | Title | Medium | Genre | Purchase Date |")
            # print the returned search list as a string for each line
            print('\n'.join((map(str, searchReturn(searchString, searchType)))))
# on change in searchbar or dropdown update output
musicSearchBar.observe(onSearchChange)  
typeDropdown.observe(onSearchChange)
```

## Music Rent - F418497 wrote on 11/12/2024

```python
'''
Selects a customer and record that can be rented and will change the Rental.txt file
and update the list when inputs are correct.

Functions:

    addNewRental(string, string)
    rent(string, string) -> boolean, boolean, boolean
    onRentButtonClicked(boolean)

Misc Variables:
    
    newRental
    addCustomer
    addedMessage
    incorrectInput
    totalCustomer
    totalRecord
    customerSearchBar 
    recordSearchBar
    rentButton
    rentOutput
'''
import subscriptionManager as sm
from datetime import date
# reads values from Subscription_Info.txt
subscriptions = sm.load_subscriptions()
def addNewRental(Record,Customer):
    '''
    Appends the record ID and customer ID to a string with today's date to a list
    and also adds this to the Rental.txt file.

        Parameters:
            Record (str): Inputted record ID string
            Customer (str): Inputted customer ID string
    '''
    newRental = []
    newRental.append(Record)
    newRental.append(date.today())
    newRental.append('')
    newRental.append(Customer)
    RentalList.append(newRental)
    newRentalStr = ','.join(str(data) for data in newRental)
    f = open("Rental.txt", "a")
    f.write('\n' + newRentalStr)
    f.close()
def rent(Customer, Record):
    '''
    This tests to see the customer ID and record ID status and will return results
    based on whether the customer doesn't exist, the record doesn't exist, the customer
    has exceeded their rental limit or if the customer is free to rent the record.

        Parameters:
            Customer (str):Inputted customer ID string
            Record (str):Inputted record ID string

        Returns:
            incorrectInput (bool):Bit output dependent on whether strings are recognised
            addCustomer (bool):Bit output dependent on whether record is available
            addedMessage (bool):Bit output dependent on whether rental limit is exceeded
    '''
    # initially all logic is assumed false
    addCustomer = False
    addedMessage = False
    incorrectInput = False
    totalCustomer, totalRecord = 0, 0
    for customer in RentalList:
        if customer[3] == Customer:
            totalCustomer += 1
        if customer[0] == Record:
            totalRecord += 1
        # if record has been rented out, but has returned, it is still available to be rented
        if customer[0] == Record and customer[2] != '': 
            addCustomer = True
        # otherwise it cannot be rented
        elif customer[0] == Record and customer[2] == '':
            addCustomer = False
    # if there is no current customer or record matching description, record cannot be rented
    if totalCustomer == 0 or totalRecord == 0:
        incorrectInput = True
    # if customer has a premium account
    if sm.check_subscription(Customer, subscriptions) == True: 
        if totalCustomer <= sm.get_rental_limit("Premium") and addCustomer == True: 
            # if customer has less records than the rental limit and the record isn't rented out, add to rental list
            addNewRental(Record,Customer)
            addedMessage = True  
    # if customer has a basic account
    else: 
        if totalCustomer <= sm.get_rental_limit("Basic") and addCustomer == True: 
            addNewRental(Record,Customer)
            addedMessage = True 
    # returns values from logic
    return incorrectInput, addCustomer, addedMessage
# create customer searchbar
customerSearchBar = widgets.Text( 
    value='',
    placeholder='Search for customer ID',
    description='Customer ID:',
    disabled = False
)
customerSearchBar.style.text_color = 'teal'
# create record searchbar
recordSearchBar = widgets.Text( 
    value='',
    placeholder='Search for record ID',
    description='Record ID:',
    disabled = False
)
recordSearchBar.style.text_color = 'teal'
# create rent button
rentButton = widgets.Button(description="Rent Record",
                           layout=widgets.Layout(width='300px', height='50px')) 
rentButton.style.button_color = 'teal'
# create rent output
rentOutput = widgets.Output(layout=widgets.Layout(height='20px', width='700px'))
def onRentButtonClicked(clicked): 
    '''
    Updates the output when the button is clicked.

        Parameters:
            clicked (bool):Boolean variable that calls the function when it is true
    '''
    rentCustomer = customerSearchBar.value
    rentRecord = recordSearchBar.value
    # start rent function when button is pressed
    incorrectInput, addCustomer, addedMessage = rent(rentCustomer, rentRecord)
    with rentOutput:
        clear_output()
        if incorrectInput == True:
            print("Customer ID or record ID is incorrect.")
        elif addCustomer == False:
            print("Record is currently rented out.")
        elif addCustomer == True and addedMessage == False:
            print("Customer has exceeded the rental limit.")
        elif addCustomer == True and addedMessage == True:
            print("Customer has been added.")
# when the button is pressed, send output
rentButton.on_click(onRentButtonClicked)
```

## Music Return - F418497 wrote on 11/12/2024

```python
'''
Returns a rented out record for an inputted customer returning a record
by giving feedback on the record and updating the rental list.

Functions:

    Return()
    onReturnButtonClicked(boolean)
    updateRecord(change)

Misc variables:

    recordsAvailable
    newReturn
    totalCustomer
    customerReturnSearchBar
    recordSelect
    ratingSlider
    feedbackSearchBar
    returnButton
    returnOutput
'''
import feedbackManager as fm
# reads values from Music_Feedback.txt
feedback_list = fm.load_feedback()
recordsAvailable = []
def Return():
    '''
    Returns a rented record by appending all of the inputted information to
    the Music_Feedback.txt file and also fills in the Rental.txt file for the
    returned record with today's date as the return day.
    '''
    returnCustomer = customerReturnSearchBar.value.strip()
    returnRecord = recordSelect.value
    rating = ratingSlider.value
    feedback = feedbackSearchBar.value
    # adds feedback to .txt file
    fm.add_feedback(returnRecord, rating, feedback, "Music_Feedback.txt")
    for customer in RentalList:
        if customer[3] == returnCustomer and customer[0] == returnRecord:
            rentalDate = customer[1]
            oldReturnStr = ','.join(str(data) for data in customer)
            # fills in return date value in Rental List
            newReturn = []
            newReturn.append(returnRecord)
            newReturn.append(rentalDate)
            newReturn.append(date.today())
            newReturn.append(returnCustomer)
            RentalList.insert(RentalList.index(customer), newReturn)
            RentalList.remove(customer)
            # removes all lines from .txt file and rewrites them with the 
            # new return date value included where the record was previously
            newReturnStr = ','.join(str(data) for data in newReturn)
            with open("Rental.txt", "r") as f:
                lines = f.readlines()
            with open("Rental.txt", "w") as f:
                for line in lines:
                    if line.strip("\n") != oldReturnStr:
                        f.write(line)
                    else:
                        f.write(newReturnStr + '\n')
            f.close()   
            break
# create customer searchbar
customerReturnSearchBar = widgets.Text( 
    value='',
    placeholder='Search for Customer',
    description='Search:',
    disabled = False,
)
customerReturnSearchBar.style.text_color = 'teal'
# create record selection
recordSelect = widgets.Select(
    options=recordsAvailable,
    value=None,
    rows=7,
    description='Record:',
    disabled=False,
    continuous_update=True
)
# create rating slider
ratingSlider = widgets.SelectionSlider(
    options=['1', '2', '3', '4', '5'],
    value='5',
    description='Rating:',
    disabled=False,
    continuous_update=False,
    orientation='horizontal',
    readout=True,
    layout=widgets.Layout(width='385px')
)
# create feedback searchbar
feedbackSearchBar = widgets.Text( # create searchbar
    value='',
    placeholder='Leave feedback',
    description='Feedback:',
    disabled = False,
)
feedbackSearchBar.style.text_color = 'teal'
# create return button
returnButton = widgets.Button(
    description='Click to return',
    disabled=False,
    icon='check',
    layout = widgets.Layout(width='300px')
)
returnButton.style.button_color = 'teal'
# create return output
returnOutput = widgets.Output(layout=widgets.Layout(height='20px', width='600px'))
def onReturnButtonClicked(clicked):
    '''
    Updates the output when button is pressed to update the user on whether 
    the record has successfully been returned.

        Parameters:
            clicked (bool):Boolean variable that calls the function when it is true
    '''
    returnCustomer = customerReturnSearchBar.value.strip()
    returnRecord = recordSelect.value
    totalCustomer = 0
    for customer in RentalList:
        if customer[3] == returnCustomer:
            totalCustomer += 1
    with returnOutput:
        clear_output()
        # if customer is invalid or no record is selected, dont return
        if totalCustomer == 0 or returnRecord == None:
            print("No customer found with that ID.")
        else:
            Return()
            print("Record returned")
def updateRecord(change):
    '''
    Updates the record select box when the customer search is changed so that
    it reads the corresponding record for the customer when a valid customer
    with a rented record is inputted.

        Parameters:
            change (bool):Boolean variable that calls the function when it is true
    '''
    returnCustomer = customerReturnSearchBar.value.strip()
    global recordsAvailable
    totalCustomer = 0
    for customer in RentalList:
        # if customer has a unreturned record(s), display on select screen
        if customer[3] == returnCustomer and customer[2] == '':
            recordsAvailable = []
            recordsAvailable.append(customer[0])
            totalCustomer += 1
    recordSelect.options = recordsAvailable
    # if there are no unreturned records, display nothing
    if totalCustomer == 0:
        recordSelect.options = []
# update record select when customer search bar is updated
customerReturnSearchBar.observe(updateRecord, names='value')
# update output when return button is pressed
returnButton.on_click(onReturnButtonClicked)
```

## Inventory Pruning

```python
'''
Creates a graph for the rental frequency of all available records and shows
the least rented records along with their average rating.
F418497 wrote on 10/12/2024

Functions:

    averageRating(list, list) -> list

Misc variables:

    allRecords
    frequency
    currentCount
    totalCount
    averageRatings
    pruningDetails
    minFrequencyRecords
    pruningGraphOutput
    pruningOutput
'''
import matplotlib.pyplot as plt
allRecords = []
frequency = []
def averageRating(frequency, ratingSum):
    '''
    Calculates the average rating of every record based on the sum of ratings 
    given to a record and the frequency of each record being rented out.

        Parameters:
            frequency (list):List of the frequency of records rented out in order
            ratingSum (list):List of the sum of rating of records rented out in order

        Returns:
            averageRatings (list):List of the average rating of records rented out in order
    '''
    averageRatings = []
    for i in range(len(ratingSum)):
        # calculate average rating for each record and round to 2dp
        averageRatings.append(round(ratingSum[i] / frequency[i], 2))
    return averageRatings
for item in RentalList:
    # append record to all records list once and add 1 to frequency list
    if allRecords.count(item[0]) == 0:
        allRecords.append(item[0])
        frequency.append('1')
    # for multiple records, override the previous frequency with the new
    # frequency and replace it in the same place it was previously
    else:
        currentCount = frequency.pop(allRecords.index(item[0]))
        totalCount = int(currentCount) + item.count(item[0])
        frequency.insert(allRecords.index(item[0]), totalCount)
# create a list with as many 0s as there are records
ratingSum = ['0'] * len(allRecords)
for item in FeedbackList:
    # add the sum of each individual record to itself and replace it from
    # where it was previously
    currentSum = int(item[1]) + int(ratingSum[allRecords.index(item[0])])
    ratingSum.pop(allRecords.index(item[0]))
    ratingSum.insert(allRecords.index(item[0]), currentSum)
# create horizontal bar chart graph
fig, ax = plt.subplots(figsize = (16,9))
bars = ax.barh(allRecords, frequency, color = 'teal')
ax.grid(color ='grey',
        linestyle ='-.', linewidth = 0.5,
        alpha = 0.5)
ax.invert_yaxis()
# set graph title and axis subtitles
ax.set_title('Record IDs and their rental frequency',
             loc ='left', fontsize=16)
plt.xlabel("Frequency",fontsize=14)
plt.ylabel("Records",fontsize=14)
minFrequencyRecords = []
for record in allRecords:
    # add all records of min frequency to a list
    if frequency[allRecords.index(record)] == min(frequency):
        minFrequencyRecords.append(record)
# create graph output
pruningGraphOutput = widgets.Output(layout=widgets.Layout(height='300px', width='512px'))
# create pruning output
pruningOutput = widgets.Output(layout=widgets.Layout(height='125px', width='700px'))
with pruningGraphOutput:
    plt.show()
with pruningOutput:
    print("Least frequently rented records:")
    print("| Record ID | Artist | Title | Medium | Genre | Purchase Date | Average Rating |")
    for music in MusicList:
        for record in minFrequencyRecords:
            if music[0] == record:
                # add all details of the least frequently rented records to a list including its average
                # rating and print it below the graph
                pruningDetails = []
                musicStr = ','.join(str(data) for data in music) + ',' + str(averageRating(frequency, ratingSum)[allRecords.index(music[0])])
                pruningDetails.append(musicStr)
                print(pruningDetails)
                pruningDetails.clear()
```
